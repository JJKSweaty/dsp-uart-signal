## Exercise 3: bits → integer (UART is LSB-first)

Implement `bits_to_byte(bits)`. If the bits are `[1, 0, 0, 0, 0, 0, 1, 0]`, the answer is `65` (ASCII `A`).

Use `value |= bit << i`, which works exactly as in C/C++.

```python
def bits_to_byte(bits):
    value = 0
    # TODO: loop through 8 bits, LSB first
    for i in range (8):
        value |= bits[i] << i
    return value
    raise NotImplementedError

assert bits_to_byte([1, 0, 0, 0, 0, 0, 1, 0]) == 65
print('ASCII:', chr(bits_to_byte([1, 0, 0, 0, 0, 0, 1, 0])))
```
