## 1. Inspect a UART capture

UART sends a low start bit, eight data bits least-significant bit first, and a high stop bit. The capture below has four samples per bit.

```python
samples_per_bit = 4
uart_samples = uart_capture(b"Hi!", samples_per_bit)
print("sample count:", len(uart_samples))
print("first 32 samples:", uart_samples[:32])
print("decoded:", decode_uart(uart_samples, samples_per_bit))
```

## 2. Decode an OOK signal

On-off keying represents a one with energy and a zero without energy. Start by averaging each group, then try center sampling.

```python
expected_bits = "101100101011"
ook_samples = repeat_bits(expected_bits, samples_per_bit=4)
print("grouped decode:", decode_grouped_bits(ook_samples, 4))
print("center decode :", decode_ook(ook_samples, 4))
```

## 3. Compare filters
